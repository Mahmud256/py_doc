```python
pay['Salary'] = pay['Annual Salary'].str.replace('$', '').str.replace(',', '').astype(float)
print(pay.head())
```

```python
print('Max Salary are: ', pay['Salary'].max())
print('Min Salary are: ', pay['Salary'].min())
print('Mean Salary are: ', pay['Salary'].mean())
```

```python
df = pay[pay['Salary'] == pay['Salary'].max()][['Name','Department','Annual Salary']]
print(df)
```

```python
df = pay[pay['Salary'] == pay['Salary'].min()][['Name','Department','Annual Salary']]
print(df)
```

```python
df = pay[pay['Salary'] < pay['Salary'].mean()].count()['Salary']
print("employee:",df)
```

```python
# Filter the DataFrame for salaries less than the mean salary
filtered_pay = pay[pay['Salary'] < pay['Salary'].mean()]

# Group the filtered DataFrame by 'Full or Part-Time' and count the 'Name' entries
name_counts = filtered_pay.groupby('Full or Part-Time')['Name'].count()

# Print the resulting counts
print(name_counts)
```

```python
# Filter the DataFrame for salaries less than the mean salary
filtered_pay = pay[pay['Salary'] < pay['Salary'].mean()]

# Group by 'Full or Part-Time' and 'Department', and then aggregate the names
grouped = filtered_pay.groupby(['Full or Part-Time', 'Department'])['Name'].count()

# Print the result
print(grouped)
```

```python
df = pay.groupby('Department')[['Salary']].mean()
print(round(df, 2))
```

```python
if 'POLICE' in df.index:
    police_salary = df.loc['POLICE']
    police_salary = round(police_salary, 2)
else:
    police_salary = 'Department POLICE not found'

print(police_salary)
```
